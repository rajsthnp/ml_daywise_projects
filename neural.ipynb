```python
inputs = [1, 2, 3, 2.5]
weights = [
    [0.2, 0.8, -0.5, 1.0],
    [0.5, -0.91, 0.26, -0.5],
    [-0.26, -0.27, 0.17, 0.87]
]
biases = [2, 3, 0.5]

output = []

for n in range(len(weights)):   # loop over each neuron
    total = 0
    for i in range(len(inputs)):
        total += inputs[i] * weights[n][i]   # multiply and sum
    total += biases[n]   # add bias for that neuron
    output.append(total)

print(output)
```

using class concenpt

```python
import numpy as np
np.random.seed(0)
X=[[1,2,3,2.5],         ##X is the input 
        [2.0,5,-1,2],
        [-1.5,2.7,3.3,-0.8]]


class layer_dense:
    def __init__(self,n_inputs, n_neurons):
        self.weights=0.10 * np.random.randn(n_inputs, n_neurons)    # matrtix of size  n_inputs * n_neurons 
        self.biases=np.zeros(1,)
    def forward(self,inputs):
        self.output=np.dot(inputs, self.weights) + self.biases

layer1 = layer_dense(4,5)
layer2 =layer_dense(5,2)

layer1.forward(X)
print(layer1.output)

layer2.forward(layer1.output)
print('\n',layer2.output)
```
